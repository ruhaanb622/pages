### Tech Talk 1: Comparisons Create Booleans

Python comparison operators return `True` or `False`.

| Python | Meaning | Example |
| --- | --- | --- |
| `==` | equal to | `inspection_score == 90` |
| `!=` | not equal to | `inspection_score != 0` |
| `>` | greater than | `inspection_score > 70` |
| `<` | less than | `inspection_score < 100` |
| `>=` | greater than or equal to | `inspection_score >= 70` |
| `<=` | less than or equal to | `inspection_score <= 100` |

````python
# CODE_RUNNER: Run the Boolean expression example

inspection_score = 82

print(inspection_score >= 70)
print(inspection_score == 100)
print(inspection_score != 0)
````

Remember:
- `inspection_score = 90` assigns a value.
- `inspection_score == 90` compares two values.

### Tech Talk 2: `and`, `or`, and `not`

A training inspection can depend on more than one Boolean value.

| Idea | Python | JavaScript | College Board pseudocode |
| --- | --- | --- | --- |
| both must be true | `and` | `&&` | `AND` |
| at least one is true | `or` | `||` | `OR` |
| reverse true/false | `not` | `!` | `NOT` |

````python
# CODE_RUNNER: Run the Boolean operators example

label_readable = True
damage_flagged = False

print(label_readable and not damage_flagged)
print(label_readable or damage_flagged)
print(not damage_flagged)
````

A longer expression can be read one part at a time. Here, a simulated item is ready for the next review only when its score is high enough **and** its label can be read:

````python
# CODE_RUNNER: Run the compound Boolean example

inspection_score = 86
label_readable = True

ready_for_review = inspection_score >= 70 and label_readable
print(ready_for_review)
````

### Tech Talk 3: Booleans Control `if` Statements

An `if` statement runs when its condition is true. In this classroom simulation, the score decides whether the item moves to the next training review station.

````python
# CODE_RUNNER: Run the if statement example

inspection_score = 72

if inspection_score >= 70:
    print("Move to the next training review station.")
else:
    print("Repeat the training inspection.")
````

The same basic decision in College Board pseudocode is:

```text
IF(inspection_score ≥ 70)
{
    DISPLAY("Move to the next training review station.")
}
ELSE
{
    DISPLAY("Repeat the training inspection.")
}
```
