### Tech Talk 1: Random Integers

Python's built-in `random` module gives us tools for random values. Here, a classroom simulator chooses one of six training stations.

```python
# CODE_RUNNER: Run the random integer example

import random

training_station = random.randint(1, 6)
print("Training station:", training_station)
```

`random.randint(a, b)` can return any whole number from `a` through `b`, including both endpoints.

For this six-station classroom simulator, the possible results are `1`, `2`, `3`, `4`, `5`, and `6`.

### Tech Talk 2: Random Choices

Python can choose directly from a list. This is useful when a training app needs to assign one item from several possible review tasks.

```python
# CODE_RUNNER: Run the random choice example

import random

inspection_tasks = [
    "helmet review",
    "suit review",
    "belt review",
    "label readability check"
]

task = random.choice(inspection_tasks)
print("Assigned training task:", task)
```

The result must be one item from the list.

The same random-integer idea looks like this in the three forms:

| Form | Random training station from 1 through 6 |
| --- | --- |
| Python | `random.randint(1, 6)` |
| JavaScript | `Math.floor(Math.random() * 6) + 1` |
| College Board pseudocode | `RANDOM(1, 6)` |

### Tech Talk 3: Use Random Values in Decisions

A program can generate a random value and then use that result in an `if` statement. This example randomly chooses which equipment review appears first in the training demo.

```python
# CODE_RUNNER: Run the random decision example

import random

first_review = random.randint(1, 2)

if first_review == 1:
    print("Start with the helmet review")
else:
    print("Start with the suit review")
```

Random does **not** mean every possible result appears once before a repeat. Repeats are normal. A training simulator may assign the same station or equipment category several times in a row.

## 3. Popcorn Hacks & Practice Tasks

### Popcorn Hack 1: Predict the Range

Before running this expression, identify every possible training-station number:

```python
# CODE_RUNNER: Run Popcorn Hack 1

import random

print(random.randint(3, 7))
```

Answer:
1. What is the smallest possible result?
2. What is the largest possible result?
3. List every possible integer.
4. Could it return `2`?
5. Could it return `7`?

Then run the code several times and compare the results with your predicted range.

### Popcorn Hack 2: Random Equipment Review

```python
# CODE_RUNNER: Run Popcorn Hack 2

import random

gear_reviews = [
    "helmet",
    "driver suit",
    "belt system",
    "window net"
]

selected_review = random.choice(gear_reviews)
print("Selected classroom review:", selected_review)
```

Do these five things:
1. List every value that can be printed.
2. Add one new motorsports equipment category.
3. Run the code at least five times.
4. Explain why the same category can appear more than once.
5. Replace the list with your own training-task category.

### Popcorn Hack 3: Fix the Station Range

This classroom simulator is supposed to choose training stations `1` through `6`, but the range is wrong.

```python
# CODE_RUNNER: Run Popcorn Hack 3

import random

station = random.randint(0, 5)
print("Training station:", station)
```

List the current possible results, fix the range so it returns `1` through `6`, and write the corrected random expression in JavaScript and College Board pseudocode.
